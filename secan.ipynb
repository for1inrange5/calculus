### **Garis Secan**

Untuk mencari kemiringan (gradien) kurva lengkung pada satu titik ternentu, $P(x, f(x))$, kita ambil titik bantuan $Q$ yang berjarak $h$ secara horizontal: $Q(x + h, f(x + h))$

Garis lurus yang memotong kedua titik ini disebut **garis secan** kemiringan garis secan dinyatakan oleh **rasio beda bagi** (difference quotient):

$$ m_{secan} = \frac{\Delta{y}}{\Delta{x}} = \frac{f(x + h) - f(x)}{(x + h) - x} = \frac{f(x + h) - f(x)}{h} $$

```python
# Mencari kecepatan rata rata

import sympy
from IPython.display import display, Math

s1, s2, t1, t2 = sympy.symbols('s1 s2 t1 t2')
nilai = {s2: 36.75, s1: 12.00, t2: 3.50, t1:2.00}

gradient = (s2 - s1) / (t2 - t1)
hasil = gradient.subs(nilai)

display(Math(sympy.latex(hasil)))
```
