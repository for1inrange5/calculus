### **Konsep Limit**

Kunci untuk menyelesaikan $\frac{0}{0}$ adalah limit, limit tidak menanyakan "berapa nilainya tepat di titik itu", melainkan "ke nilai berapa fungsi tersebut mendekat ketika input buat sedekat mungkin ke titik target"

Secara formal, notasi:
$$ \lim_{x \to c} f(x) = L $$
artinya ketika $x$ bergerak mendekati $c$ (baik dari kiri maupun kanan tetapi $x$ tidak sama dengan $c$ nilai $f(x)$ mendekati bilangan real $L$)

$$ \lim_{x \to 2} f(x) = \frac{x^2 - 4}{x - 2} $$

```python
# Dalam python, kita bisa menggunkakan sympy
from sympy import *
from IPython.display import display, Math, Latex

x = symbols('x')
f = (x**2 - 4) / (x - 2);

# Gunakan metode sympy.limit() 
limit_expr = limit(f, x, 2)

display(Math(r"\text{Fungsi} \quad f(x) = " + latex(f)))
display(Math(r"\text{hasil dari} \quad \lim_{x \to 2} f(x) = " + latex(limit_expr)))
```
